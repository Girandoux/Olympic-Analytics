```python
# =========================================================
# OLYMPIC ANALYTICS
# MACHINE LEARNING MODEL DEVELOPMENT
# =========================================================

# Projekt:
# Entwicklung eines Machine-Learning-Modells
# zur Vorhersage olympischer Medaillengewinne.
#
# Technologien:
# - pandas
# - scikit-learn
# - Random Forest
# - joblib
#
# Ziel:
# - Datenvorbereitung
# - Feature Engineering
# - Modelltraining
# - Modellspeicherung für Streamlit
#
# Output:
# - olympic_medal_model.pkl
# =========================================================
```

```python

# =========================================================
# 1. BENÖTIGTE LIBRARIES IMPORTIEREN
# =========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Machine Learning
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

# Preprocessing
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer

# Modell
from sklearn.ensemble import RandomForestClassifier

# Evaluation
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Modell speichern
import joblib
```

```python
# =========================================================
# 2. DATEN LADEN
# =========================================================

# Bereits vorbereitete ML-Datei laden

df_ml = pd.read_csv(

    "../../01_data/processed/olympics_ML_cleaned.csv"
)

print("Daten erfolgreich geladen.\n")
```

```python
# =========================================================
# 3. DATEN ÜBERPRÜFEN
# =========================================================

print(df_ml.head())

print("\nDatensatz Informationen:\n")

print(df_ml.info())

print("\nFehlende Werte:\n")

print(df_ml.isnull().sum())
```

```python
# =========================================================
# 4. FEATURES UND ZIELVARIABLE DEFINIEREN
# =========================================================

# Features für das Modell

features = [

    "age",
    "height_cm",
    "weight_kg",
    "bmi",
    "gender",
    "sport_name",
    "year",
    "country_name"
]

# Feature Matrix erstellen

X = df_ml[features]

# Zielvariable
# 1 = Medaille gewonnen
# 0 = Keine Medaille

y = df_ml["medal_win"]

print("\nFeatures und Zielvariable erstellt.")
```

```python
# =========================================================
# 5. TRAINING / TEST SPLIT
# =========================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.2,      # 20% Testdaten

    random_state=42,    # Reproduzierbarkeit

    stratify=y          # Gleichmäßige Klassenverteilung
)

print("\nTrainings- und Testdaten erstellt.")

print("X_train:", X_train.shape)

print("X_test :", X_test.shape)
```

```python
# =========================================================
# 6. NUMERISCHE UND KATEGORISCHE FEATURES DEFINIEREN
# =========================================================

# Numerische Spalten

numeric_features = [

    "age",
    "height_cm",
    "weight_kg",
    "bmi",
    "year"
]

# Kategorische Spalten

categorical_features = [

    "gender",
    "sport_name",
    "country_name"
]
```

```python
# =========================================================
# 7. PREPROCESSING FÜR NUMERISCHE FEATURES
# =========================================================

numeric_transformer = Pipeline(

    steps=[

        # Fehlende Werte ersetzen
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),

        # Numerische Werte skalieren
        (
            "scaler",
            StandardScaler()
        )
    ]
)
```

```python
# =========================================================
# 8. PREPROCESSING FÜR KATEGORISCHE FEATURES
# =========================================================

categorical_transformer = Pipeline(

    steps=[

        # Fehlende Werte ersetzen
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),

        # Textdaten encodieren
        (
            "onehot",
            OneHotEncoder(handle_unknown="ignore")
        )
    ]
)
```

```python
# =========================================================
# 9. COLUMNTRANSFORMER ERSTELLEN
# =========================================================

preprocessor = ColumnTransformer(

    transformers=[

        (
            "num",
            numeric_transformer,
            numeric_features
        ),

        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)

print("\nPreprocessing erfolgreich erstellt.")
```

```python
# =========================================================
# 10. MACHINE LEARNING MODELL DEFINIEREN
# =========================================================

# Random Forest Classifier

model = RandomForestClassifier(

    n_estimators=100,

    max_depth=10,

    random_state=42
)

print("\nRandom Forest Modell erstellt.")
```

```python
# =========================================================
# 11. ML PIPELINE ERSTELLEN
# =========================================================

pipeline = Pipeline(

    steps=[

        (
            "preprocessor",
            preprocessor
        ),

        (
            "classifier",
            model
        )
    ]
)

print("\nML Pipeline erfolgreich erstellt.")
```

```python
# =========================================================
# 12. MODELL TRAINIEREN
# =========================================================

pipeline.fit(

    X_train,
    y_train
)

print("\nModell erfolgreich trainiert.")
```

```python
# =========================================================
# 13. MODELL SPEICHERN
# =========================================================

import os

os.makedirs("models", exist_ok=True)

joblib.dump(

    pipeline,

    "../models/olympic_medal_model.pkl",

    compress=0
)

print("\nModell erfolgreich gespeichert.")

print("\nDatei erstellt:")

print("- models/olympic_medal_model.pkl")
```
